```python
import numpy as np
import matplotlib.pyplot as plt

student_scores = np.array([
    [85, 90, 65, 90],
    [60, 90, 84, 91],
    [60, 85, 55, 60],
    [98, 73, 50, 86]
])
subjects = ["Math", "Science", "English", "History"]
average_scores = np.mean(student_scores, axis=0)
print("Average marks of each subject:")
for i in range(len(subjects)):
    print(subjects[i], ":", average_scores[i])
highest = np.argmax(average_scores)
print("\nSubject with highest average score:", subjects[highest])
print("Average Score:", average_scores[highest])
plt.bar(subjects, average_scores)
plt.title("Average Marks of Each Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.show()
```

```python
### exp 22

import numpy as np
from scipy import stats

drug = np.array([10,12,8,15,11,9,13,14,10,12,11,9,15,13,12,10,14,11,13,12,9,10,14,15,11])
placebo = np.array([5,7,6,8,4,5,7,6,8,5,6,7,5,4,6,8,7,5,6,4,5,7,6,5,4])

def confidence_interval(data):
    mean = np.mean(data)
    se = stats.sem(data)
    ci = stats.t.interval(0.95, len(data)-1, loc=mean, scale=se)
    return mean, ci

drug_mean, drug_ci = confidence_interval(drug)
placebo_mean, placebo_ci = confidence_interval(placebo)

print("Drug Mean Reduction:", round(drug_mean, 2))
print("95% CI for Drug:", drug_ci)

print("\nPlacebo Mean Reduction:", round(placebo_mean, 2))
print("95% CI for Placebo:", placebo_ci)
```

```python
## exp 23

import numpy as np
from scipy.stats import ttest_ind

A = np.array([12,15,14,13,16,11,15,14,13,12])
B = np.array([18,17,16,19,15,18,17,16,19,18])

t, p = ttest_ind(A, B)

print("Mean Conversion Rate A:", np.mean(A))
print("Mean Conversion Rate B:", np.mean(B))
print("P-value:", p)

if p < 0.05:
    print("There is a statistically significant difference.")
else:
    print("There is no statistically significant difference.")
```

```python
## exp 24

import pandas as pd
import numpy as np
from scipy import stats

# Create a dummy rare_elements.csv file for demonstration
dummy_data = {
    'Concentration': np.random.rand(100) * 100 # Generating 100 random concentration values
}
dummy_df = pd.DataFrame(dummy_data)
dummy_df.to_csv('rare_elements.csv', index=False)
print("'rare_elements.csv' created successfully with dummy data.")

df = pd.read_csv("rare_elements.csv")
values = df["Concentration"].values # Changed 'data' to 'df' as 'data' was not defined before

n = int(input("Enter sample size: "))
confidence = float(input("Enter confidence level (%): "))
precision = float(input("Enter desired precision: "))

sample = values[:n]
mean = np.mean(sample)
sd = np.std(sample, ddof=1)

alpha = 1 - confidence / 100
t_value = stats.t.ppf(1 - alpha/2, n-1)
margin = t_value * sd / np.sqrt(n)

print("Point Estimate:", mean)
print("Confidence Interval:", mean-margin, "to", mean+margin)

if margin <= precision:
    print("Desired precision is achieved.")
else:
    print("More samples may be required.")
```

```python
## exp25

import pandas as pd
from scipy import stats
import numpy as np

# Create a dummy customer_reviews.csv file
dummy_review_data = {
    'Rating': [5, 4, 3, 5, 2, 4, 5, 3, 1, 5],
    'Review': [
        'Excellent product, highly recommend!',
        'Good quality, met my expectations.',
        'Average, nothing special.',
        'Very satisfied with this purchase.',
        'Disappointing, poor quality.',
        'Decent, but could be better.',
        'Love it!',
        'Okay for the price.',
        'Waste of money.',
        'Fantastic!'
    ]
}
df_dummy_reviews = pd.DataFrame(dummy_review_data)
df_dummy_reviews.to_csv('customer_reviews.csv', index=False)
print("'customer_reviews.csv' created successfully with dummy data.")

df = pd.read_csv("customer_reviews.csv")

rating = df["Rating"]
mean = rating.mean()
se = stats.sem(rating)

ci = stats.t.interval(
    0.95,
    len(rating)-1,
    loc=mean,
    scale=se
)

print("Average Rating:", round(mean, 2))
print("95% Confidence Interval:")
print("Lower:", round(ci[0], 2))
print("Upper:", round(ci[1], 2))
```

```python
### exp26

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind

control = np.array([70,72,68,75,71,69,73,74,70,72])
treatment = np.array([60,62,58,65,61,59,63,64,60,62])

t, p = ttest_ind(control, treatment)

print("Control Mean:", np.mean(control))
print("Treatment Mean:", np.mean(treatment))
print("P-value:", p)

if p < 0.05:
    print("Treatment has a statistically significant effect.")
else:
    print("Treatment has no statistically significant effect.")

plt.boxplot([control, treatment], labels=["Control", "Treatment"])
plt.ylabel("Disease Measurement")
plt.title("Control vs Treatment")
plt.show()
```

```python
### exp28
import pandas as pd
from sklearn.tree import DecisionTreeRegressor

# Create a dummy cars.csv file
dummy_car_data = {
    'Mileage': [50000, 75000, 25000, 100000, 60000, 30000, 80000, 45000, 70000, 90000],
    'Age': [5, 3, 2, 7, 4, 1, 6, 3, 5, 6],
    'Brand': [1, 2, 1, 3, 2, 1, 3, 2, 1, 3], # 1:Toyota, 2:Honda, 3:Ford
    'Engine': [1, 2, 1, 2, 1, 1, 2, 1, 2, 1], # 1:Gasoline, 2:Diesel
    'Price': [15000, 18000, 22000, 12000, 17000, 25000, 13000, 19000, 16000, 14000]
}
df_dummy_cars = pd.DataFrame(dummy_car_data)
df_dummy_cars.to_csv('cars.csv', index=False)
print("'cars.csv' created successfully with dummy data.")

df = pd.read_csv("cars.csv")

X = df[["Mileage", "Age", "Brand", "Engine"]]
y = df["Price"]

model = DecisionTreeRegressor(max_depth=4, random_state=1)
model.fit(X, y)

mileage = float(input("Enter mileage: "))
age = float(input("Enter car age: "))
brand = int(input("Enter brand code: "))
engine = int(input("Enter engine code: "))

# Convert input to DataFrame with feature names to avoid UserWarning
car_input = pd.DataFrame([[mileage, age, brand, engine]], columns=X.columns)

price = model.predict(car_input)

print("Predicted Car Price:", round(price[0], 2))

leaf = model.apply(car_input)
path = model.decision_path(car_input)

print("\nDecision Path Node IDs:")
print(path.indices)
```
